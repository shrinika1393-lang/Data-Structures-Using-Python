```python
#1. Basic For Loop
#Iterate through a list of numbers and print them.
numbers = [1, 2, 3, 4, 5]
for num in numbers:
  print(num)
```

```python
#2. Sum of Elements in a List
#Using a loop to find the total of all elements.
arr = [10, 20, 30, 40, 50]
total = 0
for val in arr:
  total += val
  print("Sum =", total)
```

```python
#3. Find Maximum Element (Array Traversal – DSA)
arr = [12, 45, 23, 51, 19, 8]
max_val = arr[0]
for num in arr:
  if num > max_val:
   max_val = num
print("Maximum Element:", max_val)
```

```python
#4. Loop Through a Dictionary
#Useful for problems involving key-value pairs (maps).
student = {"name": "Mrunal", "age": 21, "marks": 92}
for key, value in student.items():
   print(key, ":", value)
```

```python
#5. Reverse a String (Common DSA Task)
string = "Python"
reversed_str = ""
for ch in string:
   reversed_str = ch + reversed_str
   print("Reversed:", reversed_str)
```

```python
#6. Counting Frequency of Elements (DSA Hash Map Use)
arr = [1, 2, 2, 3, 1, 4, 2]
freq = {}
for num in arr:
   freq[num] = freq.get(num, 0) + 1
   print(freq)
```

```python
#7. Linear Search (Fundamental DSA Algorithm)
arr = [4, 7, 1, 9, 3]
key = 9
found = False
for val in arr:
  if val == key:
   found = True
   break
print("Element Found!" if found else "Not Found")
```

```python
#8. Factorial Using For Loop
n = 5
fact = 1
for i in range(1, n+1):
  fact *= i
  print("Factorial of", n, "is", fact)
```

```python
#9. Fibonacci Series (Iterative DSA Example)
n = 10
a, b = 0, 1
for _ in range(n):
  print(a, end=" ")
  a, b = b, a + b
```

```python
'''Simple if Statement
Executes a block only if the condition is true.'''
num = 10
if num > 0:
 print("Positive number")
```

```python
'''if-else Statement
Executes one block if true, another if false.'''
num = -5
if num >= 0:
 print("Positive number")
else:
 print("Negative number")
```

```python
'''if-elif-else Ladder
Used when there are multiple conditions to check.'''
marks = 85

if marks >= 90:
 print("Grade: A+")
elif marks >= 75:
 print("Grade: A")
elif marks >= 60:
 print("Grade: B")
else:
 print("Grade: C")
```

```python
'''Nested if-else Example
An if statement inside another if or else block.'''
num = 15

if num > 0:
  if num % 2 == 0:
   print("Positive even number")
  else:
   print("Positive odd number")
else:
 print("Number is zero or negative")
```

```python
'''DSA-Style Example: Find Largest of Three Numbers
Demonstrates nested conditional logic — similar to comparison logic in algorithms.'''
a, b, c = 10, 25, 15
if a > b:
 if a > c:
  print("Largest:", a)
 else:
  print("Largest:", c)
else:
  if b > c:
   print("Largest:", b)
  else:
   print("Largest:", c)
```

```python
'''Even–Odd and Divisibility Check
Combination of if-elif-else and nesting.'''
num = 30

if num % 2 == 0:
 if num % 5 == 0:
  print("Even and divisible by 5")
 else:
  print("Even but not divisible by 5")
else:
 print("Odd number")
```

```python
'''DSA Concept Example — Search Result
A conditional check while traversing a list.'''

arr = [3, 8, 1, 9, 5]
key = 9

found = False
for val in arr:
 if val == key:
  found = True
  break
if found:
 print("Element Found")
else:
 print("Element Not Found")
```

```python
#Nested Loop Example
for i in range(1, 4): # Outer loop for rows
  for j in range(1, 4): # Inner loop for columns
    print(j, end=" ")
    print() # Move to next line after inner loop
```

```python
#Right-Angled Triangle Pattern
# Program 1: Right-Angled Triangle
for i in range(1, 6):
  for j in range(1, i + 1):
    print("*", end=" ")
    print()
```

```python
#Inverted Right-Angled Triangle
# Program 2: Inverted Triangle

for i in range(5, 0, -1):
  for j in range(1, i + 1):
    print("*", end=" ")
print()
```

```python
#Number Triangle Pattern
# Program 3: Number Pattern
for i in range(1, 6):
  for j in range(1, i + 1):
    print(j, end=" ")
    print()
```

```python
#Pyramid Pattern
# Program 4: Pyramid Pattern
rows = 5
for i in range(1, rows + 1):
 print(" " * (rows - i), end="") # Spaces
 print("* " * i)
```

```python
#Alphabet Pattern
# Program 6: Alphabet Triangle
ch = 65 # ASCII for 'A'
for i in range(5):
  for j in range(i + 1):
   print(chr(ch), end=" ")
   ch += 1
   print()
```

```python
#Floyd’s Triangle (Number Sequence)
# Program 7: Floyd's Triangle
num = 1
for i in range(1, 6):
  for j in range(1, i + 1):
    print(num, end=" ")
    num += 1
print()
```

```python
#Diamond Pattern
# Program 8: Diamond Pattern
rows = 5
# Upper part
for i in range(1, rows + 1):
 print(" " * (rows - i) + "* " * i)
# Lower part
for i in range(rows - 1, 0, -1):
 print(" " * (rows - i) + "* " * i)
```

```python
#Continue Statement Example
for num in range(1, 11):
 if num % 3 == 0:
  continue # Skip multiples of 3
print(num)
```

```python
'''Pass Statement Example
pass is a null statement , it does nothing.
It’s used as a placeholder when a statement is syntactically required but no action is
needed yet.'''
for letter in "PYTHON":
 if letter == "H":
  pass # Placeholder — no action for 'H'
  print("Pass block executed for letter:", letter)
print("Current letter:", letter)
```
